## Data cleaning and preparation 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data/raw/Netflix.csv')
```

```python
df.columns
```

```python
critical_columns = ['show_id', 'title', 'type', 'release_year', 'duration']
df = df.dropna(subset=critical_columns)
```

```python
critical_columns = ['show_id', 'title', 'type', 'release_year', 'duration']
df = df.dropna(subset=critical_columns)
# Выбираем только строковые (object) колонки
string_columns = df.select_dtypes(include=['object']).columns

# Применяем strip() и lower() ко всем строкам в этих колонках
df[string_columns] = df[string_columns].apply(lambda x: x.str.strip().str.lower())
df
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['month_added'] = df['date_added'].dt.month
df['country'] = df['country'].fillna('Unknown')
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['listed_in'] = df['listed_in'].fillna('Unknown')
df['duration'] = df['duration'].fillna('Unknown')
```

```python
df.info()
```

```python
#Приведем типы данных в порядок
string_cols = ['show_id', 'title', 'director', 'cast', 'country', 'duration', 'listed_in', 'description']

df[string_cols] = df[string_cols].astype('string')
```

```python
df
```

```python
title_df = df[['show_id', 'title', 'type', 'cast', 'release_year', 'rating', 'duration', 'description']]

title_df['duration'] = title_df['duration'].str.extract('(\d+)').astype(int)

title_df
```

```python
directors_df = df[df['director'] != 'Unknown'][['director']].copy()

directors_df['director'] = directors_df['director'].astype('string')

df['director'] = df['director'].str.split(', ')

df = df.explode('director')

directors_df = directors_df.drop_duplicates()  # Удаление повторяющихся строк

directors_df = directors_df.reset_index(drop=True)

directors_df
```

```python
directors_df['director_id'] = [f'r{i+1}' for i in range(len(directors_df))]

directors_df['director_id'] = directors_df['director_id'].astype('string')

directors_df
```

```python
directors_df.to_csv('directors.csv', index=False)
```

```python
df1 = pd.read_csv('data/raw/Netflix.csv')

df1 = df1[['show_id', 'title', 'director', 'country', 'listed_in']]

df1['director'] = df1['director'].str.split(', ')

df1 = df1.explode('director')

df1['country'] = df1['country'].str.split(', ')

df1 = df1.explode('country')

df1['listed_in'] = df1['listed_in'].str.split(', ')

df1 = df1.explode('listed_in')


df1
```

```python
df1.to_csv('df1.csv', index=False)
```

```python
countries_df = df[df['country'] != 'Unknown'][['country']].copy()

countries_df['country'] = countries_df['country'].astype('string')

df['country'] = df['country'].str.split(', ')

df = df.explode('country')

countries_df = countries_df.drop_duplicates()  # Удаление повторяющихся строк

countries_df = countries_df.reset_index(drop=True)

countries_df
```

```python
countries_df = countries_df[countries_df['country'] != '']
countries_df
```

```python
countries_df['country_id'] = [f'с{i+1}' for i in range(len(countries_df))]

countries_df['country_id'] = countries_df['country_id'].astype('string')

countries_df
```

```python
countries_df.to_csv('countries.csv', index=True)
```

```python
genres_df = df[df['listed_in'] != 'Unknown'][['listed_in']].copy()

genres_df = genres_df.rename(columns={'listed_in': 'genre'})

genres_df['genre'] = genres_df['genre'].astype('string')

genres_df['genre'] = genres_df['genre'].str.split(', ')

genres_df = genres_df.explode('genre')

genres_df = genres_df.drop_duplicates()  # Удаление повторяющихся строк

genres_df = genres_df.reset_index(drop=True)

genres_df = genres_df[genres_df['genre'] != '']

genres_df['genre_id'] = [f'g{i+1}' for i in range(len(genres_df))]

genres_df['genre_id'] = genres_df['genre_id'].astype('string')

genres_df
```

```python
genres_df.to_csv('genres.csv', index=True)
```
